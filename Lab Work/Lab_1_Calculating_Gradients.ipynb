```python
def forward_pass(X, Y, weights, bias):
  N = X.to_numpy() @ np.array(weights)
  P = N + bias
  diff = P - np.array(Y)
  loss = (diff @ diff.T)/ (len(Y))
  # print(loss)
  return P, loss
```

```python
def backward_pass(X, Y, weights, bias, P, loss):
  Tlp = (2*np.array(P - np.array(Y)))/len(Y)
  weight_gradient = ((X.to_numpy().T) @ Tlp)
  # weights -= alpha * weight_gradient
  bias_gradient = np.sum(Tlp)
  # bias -= alpha * bias_gradient
  return (weight_gradient, bias_gradient)
```

```python
h = 1e-6
def firstDerivative(weights, b, X, Y):
  global h
  lossesH = []
  losses = []
  old_weights = weights.copy()
  new_weights = weights.copy()
  p , loss1 = forward_pass(X, Y, weights, b)
  for i in range(len(weights)):
    new_weights[i] = old_weights[i] + h
    p ,loss = forward_pass(X, Y, new_weights, b)
    losses.append(loss - loss1)
    new_weights = old_weights.copy()
  b = b+h
  p,loss = forward_pass(X,Y,weights,b)
  losses.append(loss-loss1)
  return np.array(losses)/h
```
