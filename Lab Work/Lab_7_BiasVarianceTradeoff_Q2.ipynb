```python
import numpy as np
import matplotlib.pyplot as plt
import math
from sklearn import linear_model
```

```python
def drawSamples(x):
  y = np.zeros(len(x))
  for i in range(len(x)):
    y[i] = np.sin(x[i] * math.pi)
  return y
```

```python
true_count = 1000
x_true = np.linspace(-1, 1, true_count)
y_true = np.sin(x_true * math.pi)
```

```python
def train(x_train, y_train):
  m = 0
  c = np.average(y_train)
  return m, c
```

```python
n = 2
x_train = np.random.uniform(-1, 1, n)
y_train = drawSamples(x_train)
m, c = train(x_train, y_train)
plt.ylim(-1.1, 1.1)
plt.plot(x_true, y_true, linewidth=5)
plt.scatter(x_train, y_train, color='r', marker='x')
plt.plot(x_true, c + m*x_true)
plt.show()
```

```python
runs = 100
plt.ylim(-1.1, 1.1)
plt.plot(x_true, y_true, linewidth=5)

m_s = np.zeros(runs)
c_s = np.zeros(runs)
m_avg = 0
c_avg = 0

for i in range(runs):
    n = 2
    x_train1 = np.random.uniform(-1, 1, n)
    y_train1 = drawSamples(x_train1)
    m, c = train(x_train1, y_train1)
    m_s[i] = m
    c_s[i] = c
    m_avg += m
    c_avg += c
    plt.plot(x_true, c + m*x_true, alpha=0.5)

m_avg /= runs
c_avg /= runs

plt.show()
```

```python
c_avg
```

```python
plt.ylim(-1.1, 1.1)
plt.plot(x_true, y_true, linewidth=5)
plt.plot(x_true, c_avg + m_avg * x_true, alpha=0.5)
```

```python
y_pred = c_avg + m_avg * x_true
bias = np.sum((y_pred - y_true)**2) / true_count

variance = 0
for i in range(runs):
  y_curr = c_s[i] + m_s[i]*x_true
  variance += np.sum((y_curr - y_pred)**2)
variance /= (runs * true_count)
```

```python
def train_linear(x_train, y_train):
  model = linear_model.LinearRegression()
  model.fit(x_train.reshape((-1, 1)), y_train)
  m = model.coef_[0]
  c = model.intercept_
  return m, c
```

```python
n = 2
x_train = np.random.uniform(-1, 1, n)
y_train = drawSamples(x_train)
m, c = train_linear(x_train, y_train)
plt.ylim(-1.1, 1.1)
plt.plot(x_true, y_true, linewidth=5)
plt.scatter(x_train, y_train, color='r', marker='x')
plt.plot(x_true, c + m*x_true)
plt.show()
```

```python
runs = 100
plt.ylim(-1.1, 1.1)
plt.plot(x_true, y_true, linewidth=5)

m_s = np.zeros(runs)
c_s = np.zeros(runs)
m_avg = 0
c_avg = 0

for i in range(runs):
    n = 2
    x_train1 = np.random.uniform(-1, 1, n)
    y_train1 = drawSamples(x_train1)
    m, c = train_linear(x_train1, y_train1)
    m_s[i] = m
    c_s[i] = c
    m_avg += m
    c_avg += c
    plt.plot(x_true, c + m*x_true, alpha=0.5)

m_avg /= runs
c_avg /= runs

plt.show()
```
