```python
import numpy as np
import pandas as pd
```

```python
alpha = 1e-3
epsilon = 1e-3
weightsfinal =[]
biasfinal = 0
```

```python
def prediction(x, w, b=biasfinal):
  return x@w+b
```

```python
def forward_pass(X, Y, weights, bias):
  N = X.to_numpy() @ np.array(weights)
  P = N + bias
  diff = P - np.array(Y)
  loss = (diff @ diff.T)/ (2*len(Y))
  # print(loss)
  return P, loss
```

```python
def backward_pass(X, Y, weights, bias, P, loss):
  Tlp = (np.array(P - np.array(Y)))/len(Y)
  weight_gradient = ((X.to_numpy().T) @ Tlp)
  weights -= alpha * weight_gradient
  bias_gradient = np.sum(Tlp)
  bias -= alpha * bias_gradient
  return weights, bias
```

```python
def train(X, Y):
  global weightsfinal
  global biasfinal
  weights = 0.5*np.random.rand(len(X.columns))
  bias = np.random.rand(1)
  P, loss = forward_pass(X, Y, weights, bias)
  old = loss
  while (loss > 0) :
    weights, bias = backward_pass(X, Y, weights, bias, P, loss)
    P, loss = forward_pass(X, Y, weights, bias)
    if (abs(old - loss) <= epsilon):
      break
    old = loss
  weightsfinal = weights
  biasfinal = bias
  return weights, bias
```

```python
from sklearn.model_selection import train_test_split
```

```python
X1 = 10 * np.random.randn(200)
X2 = 10 * np.random.randn(200)
y = 100 * X1 - 200 * X2 + 5
data = np.column_stack((X1, X2, y))
df = pd.DataFrame(data, columns=["x1", "x2", "y"])
X = df[["x1", "x2"]]
y = df["y"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)
```

```python
train(X_train, y_train)
```
